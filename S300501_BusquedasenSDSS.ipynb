## Ejercicio #1: ¿Cuántos shows tienen una calificación perfecta de 10.0?

```python
import sqlite3 as sql

with sql.connect("shows.db") as conn:
    cursor = conn.cursor()
    
    # Contamos las filas en ratings donde rating = 10.0
    cursor.execute("SELECT COUNT(*) FROM ratings WHERE rating = ?", (10.0,))
    count = cursor.fetchone()[0]
    print(f"1. Shows con calificación perfecta: {count}")
```

## Ejercicio #4: ¿Cuál es el show con mejor rating en el género Horror?

```python
with sql.connect("shows.db") as conn:
    cursor = conn.cursor()
    
    # Unimos shows, ratings y genres. Filtramos por género Horror y obtenemos el título
    # con el rating máximo utilizando una subconsulta para igualar el rating máximo.
    cursor.execute("""
        SELECT shows.title, ratings.rating
        FROM shows
        JOIN ratings ON shows.id = ratings.show_id
        JOIN genres ON shows.id = genres.show_id
        WHERE genres.genre = ?
          AND ratings.rating = (
              SELECT MAX(r.rating)
              FROM ratings r
              JOIN genres g ON r.show_id = g.show_id
              WHERE g.genre = ?
          )
    """, ("Horror", "Horror"))
    
    result = cursor.fetchall()
    if result:
        # Si hay más de uno con el mismo rating máximo, mostramos todos
        for row in result:
            print(f"4. Mejor show de Horror: {row[0]} (rating: {row[1]})")
    else:
        print("4. No se encontraron shows de Horror con rating.")
```

## Ejercicio #6: ¿Cuáles son los 10 shows con peor rating del género Animation entre 2005 y 2010?

```python
with sql.connect("shows.db") as conn:
    cursor = conn.cursor()
    
    # Unimos shows y ratings. Filtramos solo por año (sin filtrar por Animation).
    # Ordenamos por rating ascendente (peor primero) y limitamos a 10.
    cursor.execute("""
        SELECT shows.title, shows.year, ratings.rating
        FROM shows
        JOIN ratings ON shows.id = ratings.show_id
        WHERE shows.year BETWEEN ? AND ?
        ORDER BY ratings.rating ASC
        LIMIT 10
    """, (2005, 2010))
    
    results = cursor.fetchall()
    print("6. 10 peores shows (2005-2010):")
    for i, row in enumerate(results, 1):
        print(f"   {i}. {row[0]} ({row[1]}) - Rating: {row[2]}")
```
