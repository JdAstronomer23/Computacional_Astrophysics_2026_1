## 1. Introducción

Desde el descubrimiento de 51 Pegasi b en 1995 (Mayor & Queloz, 1995), la exoplanetología ha transitado de ser un campo de detecciones aisladas a una disciplina dominada por el *Big Data*. En la actualidad, repositorios como el *NASA Exoplanet Archive* albergan miles de sistemas planetarios confirmados (Akeson et al., 2013), abarcando una diversidad arquitectónica que reta los modelos clásicos de formación planetaria basados únicamente en nuestro Sistema Solar. 

El presente reporte detalla el desarrollo de un *pipeline* analítico *end-to-end* diseñado para caracterizar la demografía de los exoplanetas conocidos. El objetivo principal de este proyecto es interrogar programáticamente catálogos astronómicos masivos, estructurar la información en bases de datos relacionales y extraer métricas estadísticas para analizar dependencias físicas fundamentales, tales como la relación masa-radio, la distribución de temperaturas de equilibrio y los sesgos observacionales inherentes a las técnicas de detección actuales.

---

## 2. Metodología Computacional

El flujo de trabajo se estructuró en un entorno completamente reproducible basado en Python, dividiéndose en las siguientes etapas secuenciales:

*   **Adquisición de Datos:** Se utilizó la librería `Astroquery` para acceder al servicio TAP (*Table Access Protocol*) del NASA Exoplanet Archive. Se consultó específicamente la tabla `pscomppars`, aplicando filtros de calidad para descartar falsos positivos (`pl_controv_flag = 0`) y exigir valores completos en masa, radio y temperatura de equilibrio.
*   **Ingesta y Diseño de Base de Datos:** Los datos crudos se normalizaron y almacenaron en una base de datos local SQLite (`exoplanets.db`). Se implementaron índices de búsqueda y una tabla de metadatos estandarizada para garantizar la trazabilidad de la muestra.
*   **Procesamiento Analítico:** Las agregaciones estadísticas se realizaron de forma nativa mediante el motor SQL, integrando los resultados con `Pandas` para su manipulación en memoria. Se derivaron variables adicionales, como la clasificación en zonas térmicas (Fría, Templada, Caliente).
*   **Visualización y Ajustes Estadísticos:** Se utilizaron `Matplotlib`, `Seaborn` y `Scipy` para generar un *dashboard* de visualizaciones de calidad de publicación. Esto incluyó regresiones no lineales (ley de potencias) propagando el error en el espacio logarítmico, y estimaciones de densidad de kernel (KDE) para distribuciones continuas.

## 3. Resultados y Discusión

El *pipeline* extrajo y procesó exitosamente una muestra estadísticamente significativa de 5,679 exoplanetas confirmados. El análisis visual y cuantitativo de este catálogo revela tendencias fundamentales en la arquitectura planetaria y expone las severas limitaciones de nuestros instrumentos actuales.

El panel derecho detalla las frecuencias absolutas. Este sesgo instrumental favorece fuertemente el hallazgo de planetas de período corto y gran radio, estructurando las bases de datos observacionales actuales.

![Figura 2](fig2_period_radius.png)
> **Figura 2. Diagrama de dispersión bivariado del período orbital frente al radio planetario.** La muestra exhibe el clásico "Desierto Neptuniano" para períodos ultracortos ($< 3\text{ días}$) y radios intermedios ($\sim 4 R_\oplus$), producto de la intensa fotoevaporación estelar. La población de "Júpiteres Calientes" se aísla claramente en la esquina superior izquierda ($10 R_\oplus$). La escala de color mapea la temperatura de equilibrio, confirmando el decaimiento térmico en función de la distancia orbital.

![Figura 3](fig3_mass_radius.png)
> **Figura 3. Relación empírica masa-radio para la muestra observacional.** Los puntos azules representan a los exoplanetas individuales. La línea roja continua traza el mejor ajuste de la ley de potencias de la forma $R = 1.57 \cdot M^{0.35}$. La región sombreada indica el intervalo de confianza del $95\%$ propagado en el espacio logarítmico. La alta dispersión de los datos alrededor de $\sim 100 M_\oplus$ refleja la transición en la ecuación de estado degenerada y la diversidad composicional entre gigantes gaseosos y mundos rocosos masivos.

![Figura 4](fig4_temp_hist.png)
> **Figura 4. Distribución de densidad de la temperatura de equilibrio segmentada por zona térmica.** El histograma revela un profundo sesgo de selección: la inmensa mayoría de los planetas caracterizados caen en la categoría de "Calientes" ($T \ge 373\text{ K}$). Los mundos "Templados" y "Fríos" están severamente subrepresentados debido a que requieren décadas de observación continua y señales instrumentales mucho más débiles para ser detectados.

![Figura 5](fig5_spectral_box.png)
> **Figura 5. Distribución de radios planetarios en función de la clase espectral de la estrella anfitriona.** Los diagramas de violín (combinados con diagramas de caja) muestran que las estrellas tipo M (enanas rojas) albergan predominantemente planetas pequeños, con una mediana de $2.1 R_\oplus$. Por el contrario, las estrellas tipo F, G y K exhiben distribuciones más amplias y medianas superiores a $10 R_\oplus$, lo cual es consistente con la mayor disponibilidad de masa en sus discos protoplanetarios primigenios para formar gigantes gaseosos.

Se observan picos colosales en los años 2014 y 2016, marcando los hitos de validación estadística masiva de los candidatos del telescopio espacial *Kepler*. La línea negra continua demuestra un crecimiento exponencial en la última década, superando la barrera de los 5,000 planetas.

![Figura 7](fig7_corr.png)
> **Figura 7. Matriz de correlación de Pearson para parámetros físicos clave.** El mapa de calor confirma matemáticamente la Tercera Ley de Kepler mediante la correlación perfecta ($1.00$) entre el período orbital y el semieje mayor. Se evidencia un coeficiente de $0.46$ entre masa y radio. Por el contrario, el valor cercano a cero ($0.02$) entre temperatura de equilibrio y semieje mayor ilustra la limitación del coeficiente de Pearson, el cual no logra capturar la relación inversamente proporcional a la raíz cuadrada de la distancia ($T \propto a^{-1/2}$).

![Figura 8](fig8_violin.png)
> **Figura 8. Distribución de densidad de los radios planetarios segmentada por zonas de irradiación térmica.** Los diagramas de violín (con escalas logarítmicas) revelan una marcada bimodalidad en la zona "Caliente" ($T \ge 373\text{ K}$), separando físicamente a los núcleos rocosos desnudos (inferior) de los gigantes gaseosos inflados térmicamente (superior). La morfología y baja cantidad de datos en la zona "Fría" evidencian los límites de sensibilidad para detectar cuerpos menores a grandes distancias estelares.

![Figura 9](fig9_mass_temp.png)
> **Figura 9. Espacio de fase bivariado de la masa planetaria frente a la temperatura de equilibrio.** Este diagrama consolida la demografía general: los "Júpiteres Calientes" dominan la región superior derecha (altas masas y altas temperaturas). Destaca visualmente un vacío pronunciado en el centro superior (temperaturas $> 1500\text{ K}$ y masas entre $10$ y $100 M_\oplus$), corroborando la hipótesis del "Valle de la Fotoevaporación", donde los planetas gaseosos de masa intermedia pierden su envoltura debido a la extrema radiación estelar.

## 4. Conclusiones y Trabajo Futuro

El desarrollo de este *pipeline* ha permitido consolidar una visión cuantitativa de la demografía exoplanetaria actual. Físicamente, el proyecto demuestra que la distribución observada de exoplanetas es una compleja convolución entre los procesos de formación planetaria a escala global y los severos sesgos de selección de nuestra instrumentación actual. 

Desde la perspectiva computacional, la transición de flujos de análisis *ad hoc* a un entorno de programación modular y estructurado en bases de datos relacionales ha demostrado ser indispensable para el manejo riguroso de catálogos astrofísicos masivos. El tratamiento matemático cuidadoso, como el ajuste de errores en espacio logarítmico, garantiza inferencias sólidas frente a los datos.

A pesar de su robustez, el estudio asume dependencias lineales (como las evaluadas por el coeficiente de Pearson) y padece de la heterogeneidad inherente al compilar masas empíricas y estimadas. Para futuras investigaciones, se propone extender este *pipeline* incorporando modelos jerárquicos bayesianos para la clasificación probabilística de interiores planetarios y la inyección matemática de funciones de completitud observacional, permitiendo así reconstruir las verdaderas tasas de ocurrencia galáctica más allá de lo que dictan nuestros instrumentos.
